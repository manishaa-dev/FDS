```python

# Exp No: 4
# To count the frequency of occurrence of words in a body of text

# Step 1: Import required libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import gutenberg

# Step 2: Download required NLTK resources
nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')

# Step 3: Load text from the Gutenberg corpus
sample = gutenberg.raw("austen-emma.txt")

# Step 4: Tokenize the text
token = word_tokenize(sample)

# Step 5: Create a list of the first 50 tokens
wlist = []

for i in range(50):
    wlist.append(token[i])

# Step 6: Calculate the frequency of each token in the list
wordfreq = [wlist.count(w) for w in wlist]

# Step 7: Print word-frequency pairs
print("Pairs\n" + str(list(zip(wlist, wordfreq))))
```
